# Book Dataset — NumPy and Pandas

## Import statements

```python
# Import NumPy library
import numpy as np

# Import Pandas library
import pandas as pd
```

## Create a NumPy array

```python
# Create an array of book prices
prices = np.array([250, 180, 350, 120, 400, 275, 150])

# Display the original array
print("Original Book Prices:")
print(prices)
```

## Array slicing

```python
# Select elements from index 1 to index 4
print("Slice (index 1 to 4):", prices[1:5])
```

## Access the first element

```python
# Display the first book price
print("First Book Price:", prices[0])
```

## Access the last element

```python
# Display the last book price
print("Last Book Price:", prices[-1])
```

## First three elements

```python
# Display the first three book prices
print("First Three Prices:", prices[:3])
```

## Select elements from index 2 to 4

```python
# Display elements from index 2 to index 4
print("Index 2 to 4:", prices[2:5])
```

## Every second element

```python
# Display every second book price
print("Every Second Price:", prices[::2])
```

## Reverse the array

```python
# Display book prices in reverse order
print("Reverse Array:", prices[::-1])
```

## Create a two-dimensional array

```python
# Create a 2D array of book prices, ratings, and pages
books = np.array([
    [250, 4, 200],
    [180, 3, 150],
    [350, 5, 300]
])

# Display the 2D array
print("Book Information:")
print(books)
```

## Count array elements

```python
# Display the number of elements
print(prices.size)
```

## Find the maximum value

```python
# Display the highest price
print(prices.max())
```

## Find the minimum value

```python
# Display the lowest price
print(prices.min())
```

## Calculate the average

```python
# Calculate the average price
print(prices.mean())
```

## Calculate the total

```python
# Calculate the sum of all prices
print(prices.sum())
```

## Sort the array

```python
# Sort prices in ascending order
print(np.sort(prices))
```

## Create Subject Data

```python
# Create a dictionary of library details
data = {
    "Book": ["Maths Guide", "Science World", "English Grammar", "Computer Basics"],
    "Subject": ["Maths", "Science", "English", "Computer"],
    "Price": [150, 180, 120, 200]
}

# Convert dictionary into a DataFrame
data = pd.DataFrame(data)

# Display the table
print(data)

```

## Display the first 2 rows

```python
# Display the first 2 rows
print("First 2 Rows\n", data.iloc[:2])
```

## Display the last row

```python
# Display the last row
print(data.iloc[-1])
```

## Display rows from index 1 to 2

```python
# Display rows from index 1 to 2
print(data.iloc[1:3])
```

## Display the Price column

```python
# Display the Price column
print("Price Column:\n", data["Price"])
```

## Filter rows where Price is greater than 150

```python
# Display rows where Price is greater than 150
print("Rows where Price > 150:\n", data[data["Price"] > 150])
```
