# Iris Dataset — Data Analysis and Visualization

## Import libraries

```python
# Import Pandas for handling data
import pandas as pd

# Import NumPy for numerical operations
import numpy as np

# Import Matplotlib for plotting graphs
import matplotlib.pyplot as plt

# Import Seaborn for statistical visualization
import seaborn as sns
```

## Load the CSV file

```python
# Read the Iris CSV file
df = pd.read_csv("Iris.csv")

# Display the first five rows
print(df.head())
```

## Check the size of the dataset

```python
# Display the number of rows and columns
print(df.shape)
```

## Display column names

```python
# Display all column names
print(df.columns)
```

## Histogram of petal length

```python
# Plot the distribution of petal length
plt.hist(df["PetalLengthCm"], bins=5)

# Add the graph title
plt.title("Distribution of Petal Length")

# Label the X-axis
plt.xlabel("Petal Length (cm)")

# Label the Y-axis
plt.ylabel("Frequency")

# Display the graph
plt.show()
```

## Histogram of petal width using Seaborn

```python
# Plot the distribution of petal width
sns.histplot(data=df, x="PetalWidthCm", bins=15)

# Add the graph title
plt.title("Distribution of Petal Width")

# Label the X-axis
plt.xlabel("Petal Width (cm)")

# Label the Y-axis
plt.ylabel("Frequency")

# Display the graph
plt.show()
```

## Box plot of petal length

```python
# Create a box plot of petal length
plt.boxplot(df["PetalLengthCm"])

# Add the graph title
plt.title("Petal Length Box Plot")

# Label the Y-axis
plt.ylabel("Length (cm)")

# Display the graph
plt.show()
```

## Box plot of petal width using Seaborn

```python
# Create a box plot of petal width
sns.boxplot(data=df, y="PetalWidthCm")

# Add the graph title
plt.title("Petal Width Box Plot")

# Display the graph
plt.show()
```

## Compare petal length by species

```python
# Compare petal lengths of different Iris species
sns.boxplot(data=df, x="Species", y="PetalLengthCm")

# Add the graph title
plt.title("Petal Length by Iris Species")

# Display the graph
plt.show()
```

## Scatter plot of petal length and width

```python
# Plot petal length against petal width
plt.scatter(df["PetalLengthCm"], df["PetalWidthCm"])

# Add the graph title
plt.title("Petal Length vs Petal Width")

# Label the X-axis
plt.xlabel("Petal Length (cm)")

# Label the Y-axis
plt.ylabel("Petal Width (cm)")

# Display the graph
plt.show()
```

## Scatter plot by species using Seaborn

```python
# Compare petal length and width for each species
sns.scatterplot(
    data=df,
    x="PetalLengthCm",
    y="PetalWidthCm",
    hue="Species"
)

# Add the graph title
plt.title("Petal Length vs Petal Width by Species")

# Display the graph
plt.show()
```

## Pair plot

```python
# Compare relationships between numerical features
sns.pairplot(df.drop(columns=["Id"]), hue="Species")

# Display the graphs
plt.show()
```

## Correlation heatmap

```python
# Select numerical columns and exclude the ID column
numerical_df = df.select_dtypes("number").drop(
    columns=["Id"], errors="ignore"
)

# Calculate correlation between numerical features
correlation = numerical_df.corr()

# Display the correlation heatmap
sns.heatmap(correlation, annot=True, cmap="coolwarm")

# Add the graph title
plt.title("Correlation Heatmap")

# Display the graph
plt.show()
```
