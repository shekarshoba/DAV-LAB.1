## Combine Datasets Using Concatenation

```python
# Create two small dataframes from separate row ranges
df1 = df.iloc[0:5].copy()
df2 = df.iloc[5:10].copy()

# Combine the two dataframes row-wise
pd.concat([df1, df2])
```

## Merge Datasets

```python
# Create two dataframes with a shared ID
df1 = df[["gender", "math score"]].copy()
df1["ID"] = df1.index

df2 = df[["reading score"]].copy()
df2["ID"] = df2.index

# Merge using the common ID column
pd.merge(df1, df2, on="ID")
```
